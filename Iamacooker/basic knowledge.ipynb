## Sync & Async enviroment

langchain的同步与非同步
在LangChain中，同步和异步操作是通过Python的异步编程来实现的。LangChain中的许多操作都是异步的，因为它们可能需要执行耗时的任务，如与数据库通信、调用外部API或执行复杂的搜索查询。
同步操作会阻塞当前线程，直到操作完成。而异步操作不会阻塞线程，操作在后台完成，并通过回调、事件或者异步生成器与主程序交互。
